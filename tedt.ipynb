```python
p_list = []


class Product:
    index = 1 # 처음세팅 1
    def __init__(self, name, price , count):
        self.name = name # 상품명
        self.price =price #가격
        self.count = count #재고
        
        # 인덱스 세팅
        self.index= Product.index  # 부여한뒤
        Product.index += 1 # 카운트 +1
        
# 초이스
def choice():
    while True:
        print('----------------------------------------------')
        print('1.상품목록 2.담기 3.빼기 4.장바구니 5.쿠폰 6.결제 q.종료')
        print('----------------------------------------------')
        
        choice = input('선택: ')
        for choice in (1,2,3,4,5):
            return choice
        if choice == 'q':
            exit_p()
        continue
        
        
    
    
# 1. 상품목록
def product_list(p_list):
    for p in p_list:
        print(f'{p.index}. {p.name}    {p.price:,} (재고 {p.count}개)')
    
    

# 2. 담기

# 3. 뺴기

# 4. 장바구니

# 5. 쿠폰

# 6. 결제


# q = 종료
def exit_p(answer):
    if answer == 'q':
        return False
    

def main():
    # 상품 세팅
    p_list.append(Product('무선 이어폰',89_000,3))
    p_list.append(Product('텀블러',15_000,10))
    p_list.append(Product('보조 배터리',32_000,5))
    p_list.append(Product('마우스패드',9_000,0))
    p_list.append(Product('USB 허브',18_000,4))
    
    product_list(p_list)
    choice()
    # 테스트
    # print(one.name)
    # print(one.price)
    # print(one.count)
    # print(one.index)


main()

```
